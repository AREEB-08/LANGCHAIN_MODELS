```python
model = ChatOpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=os.getenv("OPENROUTER_API_KEY"),
    model="openrouter/free",
    temperature=0.7,
)
```

```python
model_multiply=model.bind_tools([multiply])
```

```python
model_multiply.invoke('can you multiply 7 and 9')
```

```python
result=model_multiply.invoke('can you multiply 7 and 9')
result
```

```python
multiply.invoke(result.tool_calls[0])
```

```python
user_query='can you multiply 7 and 8'
query=HumanMessage(user_query)
message=[query]

result = model_multiply.invoke([query])

message.append(result)

tool_result = multiply.invoke(result.tool_calls[0]["args"])
message.append(HumanMessage(str(tool_result)))

print(model_multiply.invoke(message).content)
```
